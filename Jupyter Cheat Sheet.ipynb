## Overlaying Plots

```python
plot = cw.plot([1, 2, 3]) * cw.plot([3, 2, 1])
display(plot)
```

```python
plot = cw.plot([1, 2, 3], label="plot 1").opts(color="blue")
plot *= cw.plot([3, 2, 1], label="plot 2").opts(color="red")
display(plot)
```

## Customizing Plots

```python
import holoviews as hv
plot = cw.plot()
for i in range(1, 11):
    plot *= cw.plot(range(0, i*10, i)).opts(color=hv.Palette('Spectral'))

plot.opts(title="Different Pallete", xlabel="xlabel", ylabel="ylabel")
```
